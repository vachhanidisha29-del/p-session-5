```python
import numpy as np
import pandas as pd

np.random.seed(42)

student_id = np.arange(1, 151)

df = pd.DataFrame({
    "student_id": student_id,
    "batch": np.random.choice(["A", "B", "C"], 150),
    "maths": np.random.randint(35, 101, 150),
    "science": np.random.randint(35, 101, 150),
    "english": np.random.randint(35, 101, 150),
    "attendance_pct": np.random.randint(50, 101, 150),
    "fees_paid_inr": np.random.randint(15000, 40001, 150)
})


df.loc[
    df.sample(frac=0.12, random_state=42).index,
    "maths"
] = np.nan

df.loc[
    df.sample(frac=0.08, random_state=43).index,
    "science"
] = np.nan


df.loc[
    df.sample(frac=0.10, random_state=44).index,
    "attendance_pct"
] = np.nan


df.loc[
    df.sample(n=10, random_state=45).index,
    "fees_paid_inr"
] = -1

df.to_csv("coaching_marks_raw.csv", index=False)

df.head()
```

```python

df = pd.read_csv("coaching_marks_raw.csv")

df = df.replace(-1, np.nan)

missing_report = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_pct": (df.isnull().mean() * 100).round(1)
})

missing_report
```

```python
marks_filled = df.copy()

marks_filled["maths"] = marks_filled["maths"].fillna(
    marks_filled["maths"].median()
)

marks_filled["science"] = marks_filled["science"].fillna(
    round(marks_filled["science"].mean(), 1)
)

marks_filled["attendance_pct"] = marks_filled["attendance_pct"].fillna(
    marks_filled["attendance_pct"].median()
)


marks_filled["fees_paid_inr"] = marks_filled["fees_paid_inr"].fillna(
    marks_filled["fees_paid_inr"].median()
)

marks_filled.head()
```

```python
marks_dropped = df.dropna(
    subset=["maths", "science"]
)


compare_df = pd.DataFrame({
    "version": ["raw", "dropped", "filled"],
    "row_count": [
        df.shape[0],
        marks_dropped.shape[0],
        marks_filled.shape[0]
    ],
    "mean_maths": [
        df["maths"].mean(),
        marks_dropped["maths"].mean(),
        marks_filled["maths"].mean()
    ]
})

compare_df
```
